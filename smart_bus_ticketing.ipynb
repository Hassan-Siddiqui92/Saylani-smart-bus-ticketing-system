```python
# Ticket Booking and Reservation System

# Variables Initialization

User_Choice = None
bookings = []
issued_booking_IDs = set ()
booking_counter = 1001
routes = [
    ["R101", "KARACHI", "HYDERABAD", "08:00 AM", 1200, 40, 40],
    ["R102", "KARACHI", "SUKKUR", "09:30 AM", 2000, 36, 36],
    ["R103", "HYDERABAD", "MIRPURKHAS", "11:00 AM", 800, 32, 32],
    ["R104", "KARACHI", "LAHORE", "07:00 PM", 5000, 45, 45],
    ["R105", "SUKKUR", "HYDERABAD", "02:00 PM", 1500, 30, 30]
]

# Menu

while True:
    print()
    print ('SMART BUS RESERVATION SYSTEM')
    print ('============================')
    print ('1. View Available Routes')
    print ('2. Search for a Route')
    print ('3. Book a Ticket')
    print ('4. View My Bookings')
    print ('5. Cancel a Booking')
    print ('6. View Available Seats')
    print ('7. Generate Booking Details')
    print ('8. Exit')
    print ('============================')
    print ()
    User_Choice =  input ('Enter your Choice: ').strip()
   
    # if user will select option 1
    if User_Choice == '1':
        print()
        print('=' * 100)
        print(' ' * 35, 'AVAILABLE ROUTES')
        print('=' * 100)
        
        print(f"{'Route ID':<10} {'From':<15} {'To':<18} {'Departure':<15} {'Price (Rs.)':<12} {'Total Seats':<15} {'Available':<10}")
        print('-' * 100)
        
        for route in routes:
            print(f"{route[0]:<10} {route[1]:<15} {route[2]:<18} {route[3]:<15} {route[4]:<12} {route[5]:<15} {route[6]:<10}")
        
        print('=' * 100)
            
    # if user will select option 2
    elif User_Choice == '2':
    
        print()
        print ('================')
        print ('Search for a route')
        print ('================')
        user_search = input ('Please Enter the Route ID / Departure City  / Destination City ').strip().upper()
        found = False

        for route in routes:
            if user_search == route[0] or user_search == route[1] or user_search == route[2]:

                if found == False:
                    print()
                    print('Route ID       Departure City       Destination City       Departure Time')
                    print('-----------------------------------------------------------------------')

                print(f'{route[0]:<15}{route[1]:<22}{route[2]:<23}{route[3]}')

                found = True

        if found == False:
            print('No matching route found.')
    
    # if user will select Option 3 
    elif User_Choice == '3':
        print()
        print('================')
        print('Book a Ticket')
        print('================')
        print()

        # 1. Validating passenger name
        Name = input('Please enter the name of passenger: ').strip()

        if Name == '':
            print('Passenger name cannot be blank.')
    
        else:
            Name = Name.title()
            
            # 2. Validating age
            
            age_input = input('Please enter the age: ').strip()
    
            if not age_input.isdigit() or int(age_input) <= 0:
                print('Age must be a positive integer.')
    
            else:
                age = int(age_input)
    
                # 3. Validating route ID
                route_ID = input('Please enter a valid Route ID: ').strip().upper()
    
                selected_route = None
    
                for route in routes:
                    if route[0] == route_ID:
                        selected_route = route
                        break
    
                if selected_route is None:
                    print('Error: Route ID does not exist.')
    
                else:
                    # 4. Validating  seat quantity
                    seat_input = input('Please enter number of seats to book: ').strip()
    
                    if not seat_input.isdigit() or int(seat_input) <= 0:
                        print('Invalid Seat Qauntity')
    
                    # 5. Checking Seat availability
                    elif int(seat_input) > selected_route[6]:
                        print('Not enough seats available.')
                        print('Available seats:', selected_route[6])
                    else:
                        seat_quantity = int(seat_input)
    
                        # 6. Generating unique booking ID
                        while True:
                            booking_ID = f'B{booking_counter}'
                            booking_counter += 1
    
                            if booking_ID not in issued_booking_IDs:
                                break
    
                        # 7. Calculating total fare
                        fare = selected_route[4] * seat_quantity
    
                        # 8. Storing booking
                        booking = [booking_ID,Name,age, route_ID,seat_quantity,fare,'Confirmed']
    
                        bookings.append(booking)
                        issued_booking_IDs.add(booking_ID)
    
                        # 9. Update available seats
                        selected_route[6] -= seat_quantity
    
                        # 10. Display confirmation
                        print()
                        print('=' * 40)
                        print('       BOOKING CONFIRMED!')
                        print('=' * 40)
                        print('Booking ID:', booking_ID)
                        print('Passenger:', Name)
                        print('Age:', age)
                        print('Route ID:', route_ID)
                        print('Journey:', selected_route[1], '----->',selected_route[2])
                        print('Seats booked:', seat_quantity)
                        print('Fare per seat: Rs.', selected_route[4])
                        print('Total fare: Rs.', fare)
                        print('Status: Confirmed')
                        print('=' * 40)
    
    # if user will select option 4    
    elif User_Choice == '4':
        print()
        print ('================')
        print ('Your Bookings')
        print ('================')
        if len(bookings) == 0:
            print('No bookings found.')
    
        else:
            for booking in bookings:
                print()
    
                print('=' * 40)
                print('Booking ID:', booking[0])
                print('Passenger:', booking[1])
                print('Age:', booking[2])
                print('Route ID:', booking[3])
    
                for route in routes:
                    if route[0] == booking[3]:
                        print('Journey:', route[1], '----->', route[2])
                        print('Fare per seat: Rs.', route[4])
                        break
    
                print('Seats booked:', booking[4])
                print('Total fare: Rs.', booking[5])
                print('Status:', booking[6])
                print('=' * 40)

    
    elif User_Choice == '5':
        print()
        print ('================')
        print ('Booking Cancellation')
        print ('================')
        print ()
        cancellation = input('Please enter Booking ID for cancellation: ').strip().upper()

        found = False
    
        for booking in bookings:
            if cancellation == booking[0]:
                found = True
    
                if booking[6] == 'Cancelled':
                    print('This booking has already been cancelled.')
    
                else:
                    for route in routes:
                        if route[0] == booking[3]:
                            route[6] += booking[4]
                            break
    
                    booking[6] = 'Cancelled'
    
                    print()
                    print('=' * 35)
                    print('Booking', booking[0], 'has been cancelled.')
                    print('Seats restored:', booking[4])
                    print('Route', route[0], 'available seats:', route[6])
                    print('=' * 35)
    
                break
    
        if found == False:
            print('Booking ID not found.') 
        
    elif User_Choice == '6':
        print()
        print ('================')
        print ('Available Seats')
        print ('================')
        print()
        print('1. View all routes')
        print('2. View routes with available seats')
        print()
    
        route_choice = input('Enter your choice: ').strip()
    
        if route_choice == '1' or route_choice == '2':
            print()
            print(f"{'Route ID':<12}{'From':<18}{'To':<20}{'Total Seats':<15}{'Available Seats':<18}{'Status':<20}")
            print('-' * 100)
    
            for route in routes:
    
                total_seats = route[5]
                available_seats = route[6]
    
                if available_seats < 0 or available_seats > total_seats:
                    print('Invalid seat count for route', route[0])
    
                else:
                    
                    if available_seats == total_seats:
                        status = 'Available'
    
                    elif available_seats > 0:
                        status = 'Partially Booked'
    
                    else:
                        status = 'Fully Booked'
                    
                    if route_choice == '1' or available_seats > 0:
                        print(f"{route[0]:<12}{route[1]:<18}{route[2]:<20}{total_seats:<15}{available_seats:<18}{status:<20}")
    
            print('=' * 100)

        else:
            print('Invalid choice. Please enter 1 or 2.')
    
    elif User_Choice == '7':
        total_bookings = len(bookings)
        confirmed_bookings = 0
        cancelled_bookings = 0
        seats_reserved = 0
        confirmed_revenue = 0
        fully_booked_routes = 0
        
        print()
        print ('================')
        print ('Booking Details')
        print ('================')

        for booking in bookings:
            if booking[6] == 'Confirmed':
                confirmed_bookings += 1
                seats_reserved += booking[4]
                confirmed_revenue += booking[5]

            elif booking[6] == 'Cancelled':
                cancelled_bookings += 1

        for route in routes:
            if route[6] == 0:
                fully_booked_routes += 1

        print()
        print('Total bookings created:', total_bookings)
        print('Confirmed bookings:', confirmed_bookings)
        print('Cancelled bookings:', cancelled_bookings)
        print('Seats currently reserved:', seats_reserved)
        print('Confirmed revenue: Rs.', confirmed_revenue)
        print('Fully booked routes:', fully_booked_routes)

        print()
        print('================================')
        print('  CONFIRMED BOOKINGS BY ROUTE')
        print('================================')

        for route in routes:
            route_booking_count = 0
            route_seats_reserved = 0
            route_revenue = 0

            for booking in bookings:
                if booking[3] == route[0] and booking[6] == 'Confirmed':
                    route_booking_count += 1
                    route_seats_reserved += booking[4]
                    route_revenue += booking[5]

            print()
            print('Route ID:', route[0])
            print('Journey:', route[1], '----->', route[2])
            print('Confirmed bookings:', route_booking_count)
            print('Seats reserved:', route_seats_reserved)
            print('Confirmed revenue: Rs.', route_revenue)

        print()
        print('='*40)

    elif User_Choice == '8':
        print()
        print ('='*40)
        print (' '*15 + 'Good Bye')
        print ('='*40)
        break
    else:
        print()
        print ('Enter valid input option from 1 to 8')
```
